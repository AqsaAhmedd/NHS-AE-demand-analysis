```python
import pandas as pd
import matplotlib.pyplot as plt

# Slice the dataframe to keep only Date + AE columns
# Replace 0 and 2 with the actual column indices of your Date and AE columns
monthly_clean = monthly_df.iloc[4:, [0, 2]]  # example: row 5 onward, columns 0 and 2
monthly_clean.columns = ['Date', 'AE']      # rename columns

# Convert Date to datetime, AE to numeric
monthly_clean['Date'] = pd.to_datetime(monthly_clean['Date'], errors='coerce')
monthly_clean['AE'] = pd.to_numeric(monthly_clean['AE'], errors='coerce')

# Drop any rows with missing values
monthly_clean = monthly_clean.dropna()

# Plot the trend
plt.figure(figsize=(12,6))
plt.plot(monthly_clean['Date'], monthly_clean['AE'], marker='o')
plt.title("Monthly AE Trend")
plt.xlabel("Date")
plt.ylabel("AE")
plt.grid(True)
plt.show()
```

```python
monthly_clean = monthly_df.iloc[15:, [1,2]]  # row 16 onwards, column 1 = Date, column 2 = AE
monthly_clean.columns = ['Date', 'AE']
```

```python
monthly_clean['Date'] = pd.to_datetime(monthly_clean['Date'], errors='coerce').dt.date
```

```python
monthly_clean['AE'] = pd.to_numeric(monthly_clean['AE'], errors='coerce')
monthly_clean = monthly_clean.dropna()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))
plt.plot(monthly_clean['Date'], monthly_clean['AE'], marker='o')
plt.title("Monthly A&E Attendances Trend")
plt.xlabel("Date")
plt.ylabel("AE")
plt.grid(True)
plt.show()
```

```python
monthly_clean['Rolling_Avg_6M']=monthly_clean['AE'].rolling(window=6).mean()
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))
plt.plot(monthly_clean['Date'], monthly_clean['AE'], marker='o', label='Monthly A&E')
plt.plot(monthly_clean['Date'], monthly_clean['Rolling_Avg_6M'], color='red', linewidth=2, label='6-Month Rolling Average')

plt.title("Monthly A&E Attendances Trend with Rolling Average")
plt.xlabel("Date")
plt.ylabel("A&E Attendances")
plt.legend()
plt.grid(True)
plt.show()
```

```python
monthly_clean['MoM_Growth_%'] = monthly_clean['AE'].pct_change() * 100
plt.bar(monthly_clean['Date'], monthly_clean['MoM_Growth_%'], color='orange')
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Take last 6 months rolling average as the basis
last_date = pd.to_datetime(monthly_clean['Date'].iloc[-1])
forecast_months = 12
forecast_dates = [last_date + pd.DateOffset(months=i) for i in range(1, forecast_months+1)]

# Use last rolling average as forecast for simplicity
last_avg = monthly_clean['Rolling_Avg_6M'].iloc[-1]
forecast_values = [last_avg] * forecast_months  # constant forecast

# Plot
plt.figure(figsize=(14,6))
plt.plot(monthly_clean['Date'], monthly_clean['AE'], marker='o', label='Monthly A&E')
plt.plot(monthly_clean['Date'], monthly_clean['Rolling_Avg_6M'], color='red', linewidth=2, label='6-Month Rolling Avg')
plt.plot(forecast_dates, forecast_values, color='green', linestyle='--', label='Forecast Next 12 Months')
plt.title("Monthly A&E Attendances with Rolling Average & 12-Month Forecast")
plt.xlabel("Date")
plt.ylabel("A&E Attendances")
plt.legend()
plt.grid(True)
plt.show()
```

```python
monthly_clean['YoY_Growth_%'] = monthly_clean['AE'].pct_change(periods=12) * 100
# 7️⃣ Simple 12-month forecast using last rolling average
last_date = pd.to_datetime(monthly_clean['Date'].iloc[-1])
forecast_months = 12
forecast_dates = [last_date + pd.DateOffset(months=i) for i in range(1, forecast_months+1)]
last_avg = monthly_clean['Rolling_Avg_6M'].iloc[-1]
forecast_values = [last_avg] * forecast_months

# =========================
# Plot all together
# =========================

fig, ax1 = plt.subplots(figsize=(14,6))

# Trend lines
ax1.plot(monthly_clean['Date'], monthly_clean['AE'], marker='o', label='Monthly A&E', color='blue')
ax1.plot(monthly_clean['Date'], monthly_clean['Rolling_Avg_6M'], color='red', linewidth=2, label='6-Month Rolling Avg')

ax1.set_xlabel("Date")
ax1.set_ylabel("A&E Attendances")
ax1.set_title("NHS A&E Time-Series Analysis")
ax1.grid(True)
ax1.legend(loc='upper left')

# MoM Growth bars on secondary axis
ax2 = ax1.twinx()
ax2.bar(monthly_clean['Date'], monthly_clean['MoM_Growth_%'], color='orange', alpha=0.3, label='MoM Growth (%)')
ax2.set_ylabel("MoM Growth (%)")
ax2.legend(loc='upper right')

# Optional: Add forecast
ax1.plot(forecast_dates, forecast_values, color='green', linestyle='--', label='12-Month Forecast')
ax1.legend(loc='upper left')

plt.show()

# =========================
# Display key insights (example)
# =========================
print("Key Insights:")
print("1. A&E attendances show a long-term trend with seasonal fluctuations.")
print("2. 2020 shows a sharp drop due to external factors (COVID-19).")
print("3. Rolling average reveals underlying trends clearly.")
print("4. MoM and YoY growth highlight fast increases or decreases.")
print("5. Forecast suggests stable demand over the next 12 months (based on last trend).")
```
